```python

class Vehicle:
    def __init__(self, vehicle_type):
        self.vehicle_type = vehicle_type


class Automobile(Vehicle):
    def __init__(self, vehicle_type, year, make, model, doors, roof):
        super().__init__(vehicle_type)
        self.year = year
        self.make = make
        self.model = model
        self.doors = doors
        self.roof = roof


# Get information from the user
print("Enter your car information:")

year = input("Enter the year: ")
make = input("Enter the make: ")
model = input("Enter the model: ")

# Validate the number of doors
while True:
    doors = input("Enter the number of doors (2 or 4): ")
    if doors in ["2", "4"]:
        break
    print("Please enter either 2 or 4.")

# Validate the roof type
while True:
    roof = input("Enter the roof type (solid or sun roof): ").lower()
    if roof in ["solid", "sun roof"]:
        break
    print("Please enter solid or sun roof.")

# Create an Automobile object
car = Automobile("car", year, make, model, doors, roof)

# Display the car information
print("\nVehicle Information")
print("-------------------------")
print("Vehicle type:", car.vehicle_type)
print("Year:", car.year)
print("Make:", car.make)
print("Model:", car.model)
print("Number of doors:", car.doors)
print("Type of roof:", car.roof)
```
